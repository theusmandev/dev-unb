```python
import re
from IPython.display import HTML, display

def generate_novel_post_v3(drive_url, novel_title):
    # ✅ Extract Google Drive file ID
    if "id=" in drive_url:
        file_id = drive_url.split("id=")[-1].split("&")[0]
    elif "file/d/" in drive_url:
        file_id = drive_url.split("file/d/")[1].split("/")[0]
    else:
        raise ValueError("Invalid Google Drive link")

    preview_url = f"https://drive.google.com/file/d/{file_id}/preview"
    download_url = f"https://drive.google.com/uc?export=download&id={file_id}"

    # ✅ Extract author name
    match = re.match(r"(.+?)\s+by\s+(.+)", novel_title, re.IGNORECASE)
    if match:
        novel_name = match.group(1).strip()
        author_name = match.group(2).strip()
    else:
        novel_name = novel_title
        author_name = "Famous Urdu Writer"

    # ✅ SEO Metadata
    labels = [f"{author_name} Novels", "Romantic Urdu Novels", f"{author_name}"]
    description = f"Download or read online {novel_name} by {author_name}. A heart-touching romantic Urdu novel available in PDF. Explore the best romance novels in Urdu at Urdu Novel Bank."

    # ✅ Human-Touch SEO Content
    long_description = f"""
<p><span style="font-family: Arial, sans-serif;">
If you're someone who finds peace in the pages of a soulful story, <b>{novel_name}</b> by <b>{author_name}</b> is a treat for you. In the vast world of <b>romantic novels in Urdu</b>, this particular piece stands out because of how it portrays the delicate balance of love, sacrifice, and human emotions.
</span></p>
<br>
<p><span style="font-family: Arial, sans-serif;">
At <b>Urdu Novel Bank</b>, we know the joy of discovering a <b>novel Urdu love story</b> that stays with you. <b>{novel_name}</b> is not just another book; it's a journey through the highs and lows of relationships, making it one of the <b>best romance novels in Urdu</b> to read this year.
</span></p>
<br>
<p><span style="font-family: Arial, sans-serif;">
Whether you enjoy <b>romantic novels in Urdu online reading</b> or prefer to keep a copy for later, we've got you covered. You can easily <b>free download Urdu romantic novels PDF</b> from our library. For those who love <b>Pakistani romantic novels</b> and <b>romantic stories</b>, {author_name} has delivered a narrative that feels both modern and deeply rooted in our culture.
</span></p>
<br>
<p><span style="font-family: Arial, sans-serif;">
Don't forget to check out our <b><a href="https://www.urdunovelbanks.com/search/label/Romantic%20Urdu%20Novels" target="_blank">
    Romantic Novels List
</a>
</b> for more <b>new Urdu romantic novels</b>. For now, dive into the world of {novel_name} and experience a story full of passion and loyalty.
</span></p>
<br>
"""

    # ✅ Exact Original HTML Format
    html_content = f"""<h1 style="text-align: center;">{novel_title} – Read Online &amp; Download PDF</h1>
<div>
    {long_description}
    <h2><i><span style="font-family: Arial, sans-serif;">Quick Book Overview</span></i></h2>
    <ul>
        <li><i><b>Book Title:</b> {novel_name}</i></li>
        <li><i><b>Written By:</b> {author_name}</i></li>
        <li><i><b>Language:</b> Urdu</i></li>
        <li><i><b>Availability:</b> PDF + Read Online</i></li>
    </ul><br>
    <h2><i><span style="font-family: Arial, sans-serif;">Read Online {novel_title}</span></i></h2>
</div>

<div>
    <div style="display:flex;justify-content:center;">
        <iframe loading="lazy" style="width:100%;max-width:640px;height:480px;" class="novel-iframe" src="{preview_url}"></iframe>
    </div>
    <div style="display:flex;justify-content:center;margin:20px 0;">
    </div>
    <div style="display:flex;justify-content:center;">
        <a href="{download_url}" class="custom-download-btn" target="_blank" rel="noopener noreferrer" title="Download {novel_name} By {author_name}">Download PDF</a>
    </div>
<br>
    <div style="display:flex;justify-content:center;margin-top:20px;">
        <h2 style="font-family:'Times New Roman',serif;font-style:normal;text-align:left;">
            <i><span style="font-family:Arial,sans-serif;font-weight:normal;">Novel Tags:</span></i>
        </h2>
    </div>
    <div style="display:flex;justify-content:center;">
        <i>{novel_name} PDF Free Download, {author_name} Urdu Books, Best Romantic Novel Urdu, Urdu Novel Bank, {novel_title} Online Reading</i>
    </div>
</div>
"""

    # ✅ Exact Original Colab UI Format
    display(HTML(f"""
    <h3>📄 Blogger Embed Code</h3>
    <textarea id="final_code" style="width:100%;height:300px;">{html_content}</textarea>
    <button onclick="navigator.clipboard.writeText(document.getElementById('final_code').value)">📋 Copy Embed Code</button>

    <h3>🏷️ Blogger Labels (Tags)</h3>
    <textarea id="labels_code" style="width:100%;height:100px;">{', '.join(labels)}</textarea>
    <button onclick="navigator.clipboard.writeText(document.getElementById('labels_code').value)">📋 Copy Labels</button>

    <h3>📝 SEO Description (Meta Description)</h3>
    <textarea id="description_code" style="width:100%;height:100px;">{description}</textarea>
    <button onclick="navigator.clipboard.writeText(document.getElementById('description_code').value)">📋 Copy Description</button>
    """))

# ✅ Usage
drive_link = "https://drive.google.com/file/d/1XYZabc987654321/view?usp=drive_link"
novel_title = "Bus Ulfat Si Ho Gai Hai Novel By Mahira Zainab Khan"
generate_novel_post_v3(drive_link, novel_title)
```
