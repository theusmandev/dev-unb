```python
import json
import os
import numpy as np
from sentence_transformers import SentenceTransformer

# 1. Configuration
input_filename = 'smart_urdu_novel_bank.json'
output_filename = 'smart_urdu_novel_bank_vectors.json'

# 2. Model Load
print("AI Model load ho raha hai...")
model = SentenceTransformer('all-MiniLM-L6-v2')

# 3. JSON Load aur Cleaning
if not os.path.exists(input_filename):
    print(f"Error: '{input_filename}' nahi mili!")
else:
    with open(input_filename, 'r', encoding='utf-8') as f:
        novels = json.load(f)

    print(f"Total {len(novels)} novels mile.")

    # Data Cleaning: Sirf valid titles nikalna
    cleaned_titles = []
    for novel in novels:
        title = novel.get('Titles', '')
        # Agar title string nahi hai ya khali hai, to placeholder de dein
        if title and isinstance(title, str):
            cleaned_titles.append(title)
        else:
            cleaned_titles.append("Untitled Novel")

    print("Vectors generate ho rahe hain... (Is mein 10-15 mint lag saktay hain)")

    # 4. Vectors generate karein (show_progress_bar se progress nazar ayegi)
    embeddings = model.encode(cleaned_titles, batch_size=64, show_progress_bar=True)

    # 5. Data update karein
    for i, novel in enumerate(novels):
        novel['vector'] = embeddings[i].tolist()

    # 6. Nayi file save karein
    print("File save ho rahi hai...")
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(novels, f, ensure_ascii=False)

    print(f"✅ Kaam mukammal! '{output_filename}' download ke liye taiyar hai.")
```
