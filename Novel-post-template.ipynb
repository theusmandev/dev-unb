```python
import re
from IPython.display import HTML, display

def title_case(text):
    # Capitalize first letter of each word
    return ' '.join(word.capitalize() for word in text.split())


def generate_novel_post(drive_url, novel_title):
    # ✅ Extract Google Drive file ID
    if "id=" in drive_url:
        file_id = drive_url.split("id=")[-1]
    elif "file/d/" in drive_url:
        file_id = drive_url.split("file/d/")[1].split("/")[0]
    else:
        raise ValueError("Invalid Google Drive link")

    # ✅ Create preview and download URLs
    preview_url = f"https://drive.google.com/file/d/{file_id}/preview"
    download_url = f"https://drive.google.com/uc?export=download&id={file_id}"

    # ✅ Extract novel & author name
    match = re.match(r"(.+?)\s+by\s+(.+)", novel_title, re.IGNORECASE)
    if match:
        raw_novel_name = match.group(1).strip()
        raw_author_name = match.group(2).strip()
    else:
        raw_novel_name = novel_title
        raw_author_name = "Please Write Writer Name"

    # Apply title case
    novel_name = title_case(raw_novel_name)
    author_name = title_case(raw_author_name)

    full_novel_title = f"{novel_name} By {author_name}"


    # ✅ SEO Labels
    labels = [
        "Nazia Kamran Kashif",
        "Nazia Kamran Kashif Novels"
    ]

    # ✅ SEO Meta Description (150 chars)
    description = (
        f"{full_novel_title} is a gripping Urdu novel by Nazia Kamran Kashif, "
        "famous for horror, mystery, sci-fi concepts and deep research-based storytelling."
    )[:150]

    # ✅ Long Description (Author Focused)
    long_description = f"""
<p><span style="font-family: Arial, helvetica; font-size: medium;">
<b>{novel_name}</b> is a powerful and immersive Urdu novel by <b>{author_name}</b>,
a name that stands out in contemporary Urdu literature for her
<b>horror, mystery, sci-fi, and emotionally intense storytelling</b>.
Her novels are not just stories — they are <b>experiences</b>.
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
When we talk about contemporary Urdu literature that blends <b>horror,
 mystery, and deep social emotions</b>, one name stands out: <b>Nazia Kamran Kashif</b>.
  As a dedicated mother, entrepreneur, and the visionary founder of Book Bee and PBWRP,
   she has redefined what it means to be a "Literary Activist" in Pakistan.
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
<b>Nazia Kamran Kashif</b> is a multi-talented writer, entrepreneur, mentor, and literary activist.
As a work-from-home mother and founder of <b>Book Bee</b>, she proved that the
reading culture in Pakistan is alive and thriving.
Book Bee today serves over <b>80,000+ readers</b> across Pakistan.
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
She is also the founder of <b>Pakistan’s Bloggers, Writers, Readers, and Poets (PBWRP)</b>,
a platform dedicated to mentoring and publishing new writers.
Through initiatives like <i>Human Library</i> and <i>Quote Notion</i>,
she has helped countless emerging voices find their place in Urdu literature.
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
Nazia’s life is a message to every young Pakistani:
 <b>Dream big</b>. She believes that discouragement is just another step toward motivation.
 Her mission is to ensure that every talented writer in Pakistan gets their work published, and
 through her magazine projects and online community, she is making that happen every single day.
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
What makes Nazia’s novels unforgettable is her <b>deep research</b>.
Whether she is exploring ancient graveyards, scientific mysteries,
or supernatural realities, every scene feels authentic and gripping.
Her stories keep readers hooked until the very last page.
</span></p>
<br>
<h2>Famous Novels by Nazia Kamran Kashif:</h2>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
<i>Makli (مکلی)</i> – A haunting horror novel based on Pakistan’s historic necropolis.<br>
<i>Bermuda Triangle (برمودا ٹرائی اینگل)</i> – A rare Urdu sci-fi novel blending science, Islam, and time travel.<br>
<i>Ishq Zaad (عشق زاد)</i> – A supernatural romance of obsession between a Jin-Zad and a human.<br>
<i>Dil Lagi (دل لگی)</i> – A realistic romantic journey from a simple crush to deep, selfless love.
</span></p>
<br>
<h2>Writing Style of Nazia Kamran Kashif:</h2>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
– Strong Research & Authenticity<br>
– Horror & Mystery with Psychological Depth<br>
– Emotional, Romantic & Supernatural Elements<br>
– Social Awareness & Literary Activism
</span></p>
<br>
<p><span style="font-family: Arial, helvetica; font-size: medium;">
On <b>Urdu Novel Bank</b>, you can <b>read {novel_name} online</b> or
<b>download the complete PDF</b> for free.
Support the author by purchasing physical copies from <b>Book Bee</b>.
</span></p>
"""

    # ✅ Final HTML
    html_content = f"""
<h1 style="text-align:center;">{full_novel_title} – Read Online & Download PDF</h1>

<div style="font-family: Arial, helvetica; margin:25px 0;">
{long_description}
<br>
<h2>Book Details</h2>
<div style="font-family: Arial, helvetica; font-size: medium;  margin:25px 0;">
<ul>
<li><b>Book Name:</b> {novel_name}</li>
<li><b>Author:</b> {author_name}</li>
<li><b>Status:</b> Completed</li>
<li><b>Language:</b> Urdu</li>
<li><b>Format:</b> PDF + Read Online</li>
</ul>
</div>

<h2 style="text-align:center;" >Read Online {full_novel_title}</h2>

<div style="display:flex; justify-content:center; margin:25px 0; padding:0 15px;">
  <iframe
    loading="lazy"
    src="{preview_url}"
    style="width:100%; max-width:650px; aspect-ratio:3/4; min-height:450px;">
  </iframe>
</div>

<br>

<div style="display:flex;justify-content:center; margin:25px 0; font-size: medium;">
<a href="{download_url}" target="_blank" class="custom-download-btn">Download PDF</a>
</div>
<br>
<div style="font-family: Arial, helvetica; text-align: left;">
  <blockquote>
    <h2>Connect with Writer</h2>
    <p>
      <span style="font-size: medium;"><span style="color: #ff1286;"><b>Instagram:</b></span>
      <a href="https://www.instagram.com/naziakamrank/" rel="nofollow noopener" target="_blank">
        @naziakamrank
      </a></span>
    </p>
      <p>
      <span style="font-size: medium;"><span style="color: #0766FF;"><b>Faceook:</b></span>
      <a href="https://www.facebook.com/MakliNovelwriter/" rel="nofollow noopener" target="_blank">
        @MakliNovelwriter
      </a></span>
    </p>
  </blockquote>
</div>


<div style="margin-top:25px;text-align:center;">
  <h3>Novel Tags</h3>
<i>{novel_name} PDF Free Download, Nazia Kamran Kashif Novels,
Urdu Horror Novel, Urdu Mystery Novel, Sci Fi Urdu Novel</i>
</div>
</div>
"""

    # ✅ Display Output
    display(HTML(f"""
<h3>📄 Blogger Html Code</h3>
<textarea id="final_code" style="width:100%;height:300px;">{html_content}</textarea>
<button onclick="navigator.clipboard.writeText(document.getElementById('final_code').value)">📋 Copy Embed Code</button>

<h3>🏷️ Blogger Labels (Tags)</h3>
<textarea id="labels_code" style="width:100%;height:100px;">{', '.join(labels)}</textarea>
<button onclick="navigator.clipboard.writeText(document.getElementById('labels_code').value)">📋 Copy Labels</button>

<h3>📝 SEO Description (Meta Description)</h3>
<textarea id="description_code" style="width:100%;height:100px;">{description}</textarea>
<button onclick="navigator.clipboard.writeText(document.getElementById('description_code').value)">📋 Copy Description</button>
    """))
# ✅ Example
drive_link = "https://drive.google.com/file/d/1O5KjEJ9zAt0l3YqWPY-5o9FyTrRJFeqI/view?usp=drive_link"
novel_title = "Makli novel by nazia Kamran Kashif"
generate_novel_post(drive_link, novel_title)
```
