```python
import requests
import re

def generate_safe_comment_link():
    # 1. User se URL lena
    raw_url = input("Novel ka pura URL yahan paste karen: ")
    blog_id = "6060373960762107183"  # Aapki Blog ID

    # 2. URL ko Clean karna (? aur # se pehle wala hissa lena)
    clean_url = raw_url.split('?')[0].split('#')[0]
    print(f"\n[1] Cleaned URL: {clean_url}")

    # 3. Website ka Source Code fetch karna
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }

    try:
        response = requests.get(clean_url, headers=headers)
        response.raise_for_status()
        source_code = response.text

        # 4. Post ID search karna (Regex ke zariye)
        # Blogger source mein aksar 'postId': 'number' hota hai
        match = re.search(r"['\"]postId['\"]\s*:\s*['\"](\d+)['\"]", source_code)

        if match:
            post_id = match.group(1)
            print(f"[2] Post ID Found: {post_id}")

            # 5. Final Safe Comment Link banana
            final_link = f"https://www.blogger.com/comment/fullpage/post/{blog_id}/{post_id}"

            print("\n" + "="*50)
            print("AAPKA SAFE COMMENT LINK TAIYAR HAI:")
            print(final_link)
            print("="*50)

            # Colab mein clickable link display karna
            from IPython.display import display, HTML
            display(HTML(f'<a href="{final_link}" target="_blank">Click Here to Open Comment Box</a>'))

        else:
            print("[-] Error: Source code mein Post ID nahi mil saki.")

    except Exception as e:
        print(f"[-] Error: Website load nahi ho saki. {e}")

# Function run karen
generate_safe_comment_link()
```
